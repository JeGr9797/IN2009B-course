### Capacitated Facility Location Problem

Let $I = \{1, \ldots, i, \ldots, m\}$ be the index set of potential facility locations, and let $J = \{1, \ldots, j, \ldots, n\}$ be the index set of customers. We refer to potential facility locations by their indices; therefore, we say that facility $i$ is open if the decision is made to establish a service center at potential location $i$. Likewise, customers are denoted by their indices, and we simply refer to customer $j$. Associated with each $i \in I$, let $q_i$ denote the maximum capacity of facility $i$. The service demand of customer $j$ is denoted by $d_j$. There are two types of costs. The decision to establish a facility at $i \in I$ incurs a fixed opening cost $f_i$. For $i \in I$ and $j \in J$, $c_{ij}$ denotes the cost of serving the entire demand of customer $j$ from facility $i$.

Classical formulations for the FLP use two sets of decision variables: one for selecting which facilities to open and another for assigning customer demand to the open facilities. For the location decisions, for each $i \in I$ we define:

$$
y_i =
\begin{cases}
1 & \text{if a facility is open at location } i, \\
0 & \text{otherwise.}
\end{cases}
$$

For the assignment decisions, for each $i \in I$ and $j \in J$ we define:

$$
x_{ij} =
\begin{cases}
1 & \text{if the demand of customer } j \text{ is served by facility } i, \\
0 & \text{otherwise.}
\end{cases}
$$

A standard integer programming formulation for the FLP is as follows:

$$
\begin{aligned}
\text{min} \quad & \sum_{i \in I} f_i y_i + \sum_{i \in I} \sum_{j \in J} c_{ij} x_{ij} \\
\text{subject to} \quad & \sum_{i \in I} x_{ij} = d_j \quad \forall j \in J \\
& \sum_{j \in J} x_{ij} \leq q_i y_i \quad \forall i \in I \\
& y_i \in \{0,1\} \quad \forall i \in I \\
& x_{ij} \in \{0,1\} \quad \forall i \in I,\; \forall j \in J
\end{aligned}
$$


```python
%pip install gurobipy
from gurobipy import Model, GRB, quicksum
import numpy as np
```

```python
# Number of facilities
facilities = 5

# Number of customers
customers = 5

# Sets
I = range(facilities)
J = range(customers)

# Symmetric shipping cost matrix
c = np.array([[0, 3, 3, 6, 3],
              [3, 0, 4, 5, 5],
              [3, 4, 0, 2, 4],
              [6, 5, 2, 0, 5],
              [3, 5, 4, 5, 0]])

# Facility opening cost
f = np.array([25.0, 25.0, 25.0, 25.0, 25.0])

# Demand
d = np.array([10, 8, 5, 8, 12])

# Capacity
q = np.array([15, 15, 15, 15, 15])
```

```python
# Define model
m = Model("fixed-charge")

# Decision variables
#x = m.addVars(I, J, vtype=GRB.BINARY, name="x")
x = m.addVars(I, J, name="x")
y = m.addVars(I, vtype=GRB.BINARY, name="y")

# Objective function
m.setObjective(quicksum(f[i]*y[i] for i in I) + quicksum(c[i,j]*x[i,j]*d[j] for i in I for j in J), GRB.MINIMIZE)

# Constraints
m.addConstrs(quicksum(x[i,j] for i in I) == 1 for j in J)
m.addConstrs(quicksum(x[i,j]*d[j] for j in J) <= q[i]*y[i] for i in I)

# Optimize
m.optimize()
```

```python
if m.status == GRB.OPTIMAL:
    print("Optimal solution")
    print("Total cost: ", m.objVal)
    for i in I:
      if y[i].X > 0.1:
          print("Open facility: ", i)
else:
    print("There is not optimal solution")
```
